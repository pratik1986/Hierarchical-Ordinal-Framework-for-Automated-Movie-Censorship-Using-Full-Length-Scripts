```python
import numpy as np
import pandas as pd
import torch
import os
import joblib # Import joblib for loading thresholds
from sklearn.metrics import classification_report, confusion_matrix, cohen_kappa_score, mean_absolute_error
from scipy.stats import spearmanr
from sklearn.metrics import mean_absolute_error, mean_squared_error, accuracy_score, precision_recall_fscore_support
from transformers import BertModel, BertTokenizer # Import necessary components if not available
import torch.nn as nn # Import nn module
from sklearn.model_selection import train_test_split # Import train_test_split
import re # Import re for preprocess_text

# Re-define classes and functions needed for model loading and evaluation
# Assuming OrdinalRegressionLoss, AttentionPooling, HierarchicalBERTOrdinal,
# ScriptDataset, ScriptDataCollator are defined in previous cells.
# Re-defining them here for self-containment in this step's execution.

# Define a custom loss function for ordinal regression
class OrdinalRegressionLoss(nn.Module):
    def __init__(self, num_labels, class_weights=None):
        super(OrdinalRegressionLoss, self).__init__()
        self.num_labels = num_labels
        self.class_weights = class_weights

    def forward(self, cumulative_logits, labels):
        batch_size = labels.size(0)
        cumulative_targets = torch.zeros(batch_size, self.num_labels - 1, dtype=torch.float32).to(cumulative_logits.device)
        for i in range(batch_size):
            if labels[i] > 0:
                cumulative_targets[i, :labels[i]] = 1.0

        loss = nn.functional.binary_cross_entropy_with_logits(cumulative_logits, cumulative_targets, reduction='mean')
        return loss

# Define Attention Pooling module
class AttentionPooling(nn.Module):
    def __init__(self, hidden_size):
        super(AttentionPooling, self).__init__()
        self.attention_layer = nn.Linear(hidden_size, 1)
        self.softmax = nn.Softmax(dim=1)

    def forward(self, chunk_embeddings):
        attention_scores = self.attention_layer(chunk_embeddings).squeeze(2)
        attention_weights = self.softmax(attention_scores).unsqueeze(2)
        document_embedding = torch.sum(chunk_embeddings * attention_weights, dim=1)
        return document_embedding

class HierarchicalBERTOrdinal(nn.Module):
    def __init__(self, num_labels, bert_model_name='bert-base-uncased', freeze_layers=6):
        super(HierarchicalBERTOrdinal, self).__init__()
        self.num_labels = num_labels
        self.freeze_layers = freeze_layers

        self.chunk_encoder = BertModel.from_pretrained(bert_model_name)

        # Freezing layers is part of the model definition but doesn't need to be applied again if loading state_dict
        # The state_dict contains the learned weights, but the requires_grad settings are part of the model structure definition.
        # Ensure requires_grad is set correctly if you intend to potentially continue training or analyze gradients.
        # For simple inference/evaluation, requires_grad=False is fine for frozen layers.
        for param in self.chunk_encoder.embeddings.parameters():
             param.requires_grad = False

        for i in range(self.freeze_layers):
            for param in self.chunk_encoder.encoder.layer[i].parameters():
                param.requires_grad = False


        self.attention_pooling = AttentionPooling(self.chunk_encoder.config.hidden_size)
        self.ordinal_layer = nn.Linear(self.chunk_encoder.config.hidden_size, num_labels - 1)
        # The loss function is not needed for inference/evaluation

    def forward(self, input_ids, attention_mask, token_type_ids=None, labels=None):
        chunk_embeddings = []
        # Get the device of the model's first parameter (or any parameter) to ensure consistency
        model_device = next(self.parameters()).device

        for i in range(len(input_ids)):
            # Move input tensors to the model's device
            chunk_input_ids = input_ids[i].to(model_device)
            chunk_attention_mask = attention_mask[i].to(model_device)
            chunk_token_type_ids = token_type_ids[i].to(model_device) if token_type_ids is not None and token_type_ids[i] is not None else None

            # Ensure tensors have a batch dimension
            chunk_input_ids = chunk_input_ids.unsqueeze(0)
            chunk_attention_mask = chunk_attention_mask.unsqueeze(0)
            if chunk_token_type_ids is not None:
                chunk_token_type_ids = chunk_token_type_ids.unsqueeze(0)

            outputs = self.chunk_encoder(input_ids=chunk_input_ids,
                                         attention_mask=chunk_attention_mask,
                                         token_type_ids=chunk_token_type_ids)

            # Use the [CLS] token embedding as the chunk representation and move to the model's device
            chunk_embedding = outputs.last_hidden_state[:, 0, :].to(model_device) # Explicitly move to model_device
            chunk_embeddings.append(chunk_embedding)


        # Stack chunk embeddings to form a sequence for the attention pooling
        # Ensure chunk_embeddings_tensor is on the correct device (model_device)
        chunk_embeddings_tensor = torch.cat(chunk_embeddings, dim=0).unsqueeze(0).to(model_device)


        document_embedding = self.attention_pooling(chunk_embeddings_tensor)

        cumulative_logits = self.ordinal_layer(document_embedding)

        # Return logits only for evaluation
        return cumulative_logits

# Define Dataset and Data Collator classes (assuming defined in previous cells)
class ScriptDataset(torch.utils.data.Dataset):
    def __init__(self, dataframe):
        self.dataframe = dataframe

    def __len__(self):
        return len(self.dataframe)

    def __getitem__(self, idx):
        row = self.dataframe.iloc[idx]
        return {
            'tokenized_chunks': row['tokenized_chunks'],
            'labels': row['rating_encoded'] # Include labels for evaluation
        }

class ScriptDataCollator:
    def __call__(self, features):
        # Assuming features contains a list of dictionaries, each with 'tokenized_chunks' and 'labels'
        # Each 'tokenized_chunks' is a list of tokenized chunk outputs (from tokenizer)

        input_ids_list = [item['tokenized_chunks'] for item in features] # List of lists of tokenized tensors
        labels = torch.tensor([item['labels'] for item in features], dtype=torch.long) # Shape: (batch_size,)

        # For our model which processes one script (a list of chunks) at a time in the forward pass,
        # the data collator needs to return the list of tokenized chunk outputs directly.
        # We also need to return the labels.

        # Assuming batch size is 1 for evaluation as well based on previous Trainer args
        # If batch size is > 1, the forward method needs to be adjusted to handle batches of scripts.
        # With batch_size=1, len(features) is 1.
        if len(features) != 1:
             raise ValueError("Batch size must be 1 for this data collator and model forward pass.")

        tokenized_chunks = features[0]['tokenized_chunks'] # Get the list of tokenized chunks for the single script

        input_ids = [chunk['input_ids'].squeeze(0) for chunk in tokenized_chunks] # List of (seq_len,) tensors
        attention_mask = [chunk['attention_mask'].squeeze(0) for chunk in tokenized_chunks] # List of (seq_len,) tensors
        token_type_ids = [chunk['token_type_ids'].squeeze(0) if 'token_type_ids' in chunk else None for chunk in tokenized_chunks] # List of (seq_len,) tensors or None


        return {
            'input_ids': input_ids,
            'attention_mask': attention_mask,
            'token_type_ids': token_type_ids if all(t is not None for t in token_type_ids) else None,
            'labels': labels.squeeze(0) # Labels should be a single scalar for batch size 1
        }


# --- Data Preparation (Including necessary steps to ensure dataframes are available) ---
# Assuming base_path is defined in a previous cell. If not, define it here.
base_path = "/content/drive/MyDrive/PhDWorks/4_Final_Writing_Papers/05_ConferencePaper_3/dataset/English_unbal"

# Assuming parse_filename function is defined in a previous cell. If not, define it here.
def parse_filename(filename):
    """Robust filename parser that handles titles with 'The' and other edge cases"""
    basename = os.path.splitext(filename)[0]
    parts = basename.split('_')

    # Normalize ratings
    rating_map = {
        'PG13': 'PG-13',
        'PG-13': 'PG-13',
        'NC17': 'NC-17',
        'NC-17': 'NC-17'
    }
    rating = parts[0].upper()
    rating = rating_map.get(rating, rating)

    # Improved year extraction using regex to find 4-digit numbers
    year = None
    year_pattern = re.compile(r'^(19|20)\d{2}$')  # Only 1900-2099

    # Check from end backwards first
    for part in reversed(parts):
        if year_pattern.match(part):
            year = part
            break

    # If not found, check any position
    if not year:
        for part in parts:
            if year_pattern.match(part):
                year = part
                break

    if not year:
        raise ValueError(f"Year not found in filename: {filename}")

    # Get all parts between rating and year
    year_idx = parts.index(year)
    title_parts = parts[1:year_idx]

    # Special handling for titles containing "The"
    if len(title_parts) > 1 and title_parts[-1] == 'The':
        # Move "The" to beginning (e.g., ["Matrix", "The"] -> ["The", "Matrix"])
        title_parts = [title_parts[-1]] + title_parts[:-1]

    title = ' '.join(title_parts)  # Use space instead of underscore for readability

    return {
        'filename': filename,
        'rating': rating,
        'title': title,  # Now with spaces instead of underscores
        'year': int(year),
        'filepath': os.path.join(base_path, filename)
    }

# Create metadata dataframe
metadata = []
for filename in os.listdir(base_path):
    if filename.endswith('.txt'):
        try:
            metadata.append(parse_filename(filename))
        except Exception as e:
            print(f"Error processing {filename}: {str(e)}")

df = pd.DataFrame(metadata)


# Define the desired ordinal mapping (assuming defined in previous cells)
rating_mapping = {
    'G': 0,
    'PG': 1,
    'PG-13': 2,
    'R': 3,
    'NC-17': 4
}
# Map ordinal labels to numerical values
df['rating_encoded'] = df['rating'].map(rating_mapping)

# Create a mapping from encoded labels back to original ratings for clarity
encoded_to_rating = {v: k for k, v in rating_mapping.items()}


# Set a global seed for reproducibility (assuming defined in previous cells)
GLOBAL_SEED = 42
# from transformers import set_seed # Import set_seed if not available
# set_seed(GLOBAL_SEED) # Set seed for reproducibility

# Split data into training, validation, and test sets
train_df, test_df = train_test_split(df, test_size=0.2, random_state=GLOBAL_SEED, stratify=df['rating_encoded'])
train_df, val_df = train_test_split(train_df, test_size=0.25, random_state=GLOBAL_SEED, stratify=train_df['rating_encoded']) # 0.25 of 0.8 is 0.2

# Define text preprocessing function (assuming defined in previous cells)
def preprocess_text(filepath):
    """Loads and preprocesses text from a file."""
    with open(filepath, 'r', encoding='utf-8', errors='ignore') as f:
        text = f.read()

    # Basic cleaning: lowercasing, remove punctuation and special characters, normalize whitespace
    text = text.lower()
    text = re.sub(r'[^a-z0-9\s]', '', text) # Keep letters, numbers, and whitespace
    text = re.sub(r'\s+', ' ', text).strip() # Normalize whitespace

    return text

# Define a maximum chunk size and overlap (assuming defined in previous cells)
MAX_CHUNK_SIZE = 512
OVERLAP = 50

def chunk_script(script, chunk_size=MAX_CHUNK_SIZE, overlap=OVERLAP):
    """Splits a script into overlapping chunks."""
    words = script.split()
    chunks = []
    # Calculate step size, ensuring it's at least 1
    step = max(1, chunk_size - overlap)
    for i in range(0, len(words), step):
        # Ensure the last chunk includes the end of the script
        chunk = words[i:i + chunk_size]
        chunks.append(" ".join(chunk))
    return chunks

# Initialize BERT tokenizer (assuming defined in previous cells)
tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')

def tokenize_chunks(chunks):
    """Tokenizes a list of text chunks."""
    return [tokenizer(chunk, truncation=True, padding='max_length', max_length=MAX_CHUNK_SIZE, return_tensors='pt') for chunk in chunks]


# Load and preprocess text for each file and add to the split dataframes
train_df['clean_script'] = train_df['filepath'].apply(preprocess_text)
val_df['clean_script'] = val_df['filepath'].apply(preprocess_text)
test_df['clean_script'] = test_df['filepath'].apply(preprocess_text)

# Apply chunking to the 'clean_script' column for the split dataframes
train_df['script_chunks'] = train_df['clean_script'].apply(lambda x: chunk_script(x, chunk_size=MAX_CHUNK_SIZE, overlap=OVERLAP))
val_df['script_chunks'] = val_df['clean_script'].apply(lambda x: chunk_script(x, chunk_size=MAX_CHUNK_SIZE, overlap=OVERLAP))
test_df['script_chunks'] = test_df['clean_script'].apply(lambda x: chunk_script(x, chunk_size=MAX_CHUNK_SIZE, overlap=OVERLAP))

# Apply tokenization
train_df['tokenized_chunks'] = train_df['script_chunks'].apply(tokenize_chunks)
val_df['tokenized_chunks'] = val_df['script_chunks'].apply(tokenize_chunks)
test_df['tokenized_chunks'] = test_df['script_chunks'].apply(tokenize_chunks)


# --- Configuration ---
# Assuming num_labels, encoded_to_rating, test_df, val_df are available
# Define num_labels and encoded_to_rating if not available
num_labels = len(rating_mapping)


# Define the path to the saved model state dictionary
model_output_dir = "/content/drive/MyDrive/PhDWorks/4_Final_Writing_Papers/05_ConferencePaper_3/dataset/outputs/English_Bert_ordinal_unbal/"
model_path = os.path.join(model_output_dir, "hierarchical_bert_ordinal_attention_model.pth")

# Define the number of frozen layers used during training
FREEZE_LAYERS = 6 # Assuming 6 layers were frozen

# --- Model Initialization and Loading ---
print("Initializing model and loading state dictionary...")
model = HierarchicalBERTOrdinal(num_labels=num_labels, freeze_layers=FREEZE_LAYERS)

# Check if a GPU is available and move the model to GPU
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model.to(device)

# Load the saved state dictionary
if os.path.exists(model_path):
    model.load_state_dict(torch.load(model_path, map_location=device))
    print(f"Model state dictionary loaded from: {model_path}")
else:
    print(f"Error: Model state dictionary not found at {model_path}")
    # Exit or handle the error appropriately
    exit() # Exit the cell execution if model file is not found

model.eval() # Set the model to evaluation mode

# --- Prepare Data for Prediction and Calibration ---
# Assuming test_df and val_df are available from previous data loading steps.
# If not, re-load and preprocess them here.
# Based on the notebook state, these DataFrames are available and populated with 'tokenized_chunks'.

# Create datasets
test_dataset = ScriptDataset(test_df)
val_dataset = ScriptDataset(val_df)

# Create data collator
data_collator = ScriptDataCollator()

# Use a simple DataLoader for prediction as Trainer is not strictly needed for inference
from torch.utils.data import DataLoader

test_dataloader = DataLoader(test_dataset, batch_size=1, collate_fn=data_collator)
val_dataloader = DataLoader(val_dataset, batch_size=1, collate_fn=data_collator)


# --- 1. Perform Ordinal Calibration ---

# Make predictions on the validation set
print("\nMaking predictions on the validation set for calibration...")
val_cumulative_logits = []
val_true_labels = []

with torch.no_grad():
    for batch in val_dataloader:
        # Move batch data to the same device as the model
        # Ensure each tensor within the batch is moved to the device
        batch_input_ids = [t.to(device) for t in batch['input_ids']]
        batch_attention_mask = [t.to(device) for t in batch['attention_mask']]
        batch_token_type_ids = [t.to(device) if t is not None else None for t in batch['token_type_ids']] if batch['token_type_ids'] is not None else None
        batch_labels = batch['labels'].to(device)


        # The forward method of HierarchicalBERTOrdinal expects lists of tensors for input_ids, etc.
        # The data collator returns these lists for a batch size of 1.

        logits = model(input_ids=batch_input_ids, attention_mask=batch_attention_mask, token_type_ids=batch_token_type_ids)

        val_cumulative_logits.append(logits.squeeze(0).cpu().numpy()) # Remove batch dim, move to cpu, convert to numpy
        val_true_labels.append(batch_labels.item()) # Get the scalar label

val_cumulative_logits = np.vstack(val_cumulative_logits) # Stack logits from all validation samples
val_true_labels = np.array(val_true_labels)

print(f"Validation predictions shape: {val_cumulative_logits.shape}")
print(f"Validation true labels shape: {val_true_labels.shape}")

# Find optimal thresholds on the validation set
print("\nSearching for optimal ordinal thresholds on the validation set...")

num_threshold_steps = 20
best_qwk = -1.0
best_thresholds = np.zeros(num_labels - 1)

min_logit = np.min(val_cumulative_logits)
max_logit = np.max(val_cumulative_logits)

# Ensure a reasonable range even if min/max are close
search_range_min = min(min_logit - 1.0, -5.0)
search_range_max = max(max_logit + 1.0, 5.0)

possible_threshold_values = [np.linspace(search_range_min, search_range_max, num_threshold_steps) for _ in range(num_labels - 1)]

from itertools import product

for thresholds in product(*possible_threshold_values):
    thresholds_np = np.array(thresholds)

    current_predicted_classes = np.sum(val_cumulative_logits > thresholds_np, axis=1)
    current_predicted_classes = np.clip(current_predicted_classes, 0, num_labels - 1)

    current_qwk = cohen_kappa_score(val_true_labels, current_predicted_classes, weights='quadratic')

    if current_qwk > best_qwk:
        best_qwk = current_qwk
        best_thresholds = thresholds_np

print(f"\nOptimal thresholds found on validation set (maximizing QWK): {best_thresholds}")
print(f"Best QWK on validation set with these thresholds: {best_qwk:.4f}")

# Save the calibrated thresholds
thresholds_path = os.path.join(model_output_dir, 'calibrated_thresholds.pkl')
joblib.dump(best_thresholds, thresholds_path)
print(f"\nCalibrated thresholds saved to: {thresholds_path}")


# Apply the learned thresholds to the test set predictions
print("\nMaking predictions on the test set...")
test_cumulative_logits = []
test_true_labels = []

with torch.no_grad():
     for batch in test_dataloader:
        # Move batch data to the same device as the model
        # Ensure each tensor within the batch is moved to the device
        batch_input_ids = [t.to(device) for t in batch['input_ids']]
        batch_attention_mask = [t.to(device) for t in batch['attention_mask']]
        batch_token_type_ids = [t.to(device) if t is not None else None for t in batch['token_type_ids']] if batch['token_type_ids'] is not None else None
        batch_labels = batch['labels'].to(device)


        logits = model(input_ids=batch_input_ids, attention_mask=batch_attention_mask, token_type_ids=batch_token_type_ids)

        test_cumulative_logits.append(logits.squeeze(0).cpu().numpy()) # Remove batch dim, move to cpu, convert to numpy
        test_true_labels.append(batch_labels.item()) # Get the scalar label

test_cumulative_logits = np.vstack(test_cumulative_logits)
test_true_labels = np.array(test_true_labels)


# Apply the learned optimal thresholds to the test set cumulative logits
calibrated_test_predicted_classes = np.sum(test_cumulative_logits > best_thresholds, axis=1)
calibrated_test_predicted_classes = np.clip(calibrated_test_predicted_classes, 0, num_labels - 1)

print("Test set predictions obtained and calibrated.")

# --- 2. Calculate and Report Calibrated Test Metrics ---

print("\n--- Reporting Metrics for Calibrated Test Set ---")

# Ensure class_labels is defined (assuming encoded_to_rating is available)
class_labels = list(encoded_to_rating.values())

# Calculate metrics for the calibrated test set
calibrated_test_accuracy = accuracy_score(test_true_labels, calibrated_test_predicted_classes)
calibrated_test_mae = mean_absolute_error(test_true_labels, calibrated_test_predicted_classes)
calibrated_test_quadratic_kappa = cohen_kappa_score(test_true_labels, calibrated_test_predicted_classes, weights='quadratic')
try:
    calibrated_test_spearman_rho, _ = spearmanr(test_true_labels, calibrated_test_predicted_classes)
except ValueError:
    calibrated_test_spearman_rho = np.nan # Handle case where one array is constant

calibrated_report = classification_report(test_true_labels, calibrated_test_predicted_classes, target_names=class_labels, zero_division=0, output_dict=True)
calibrated_test_macro_f1 = calibrated_report['macro avg']['f1-score']

calibrated_test_cm = confusion_matrix(test_true_labels, calibrated_test_predicted_classes, labels=np.arange(num_labels))


# Overall Metrics
print(f"Accuracy: {calibrated_test_accuracy:.4f}")
print(f"Macro F1-Score: {calibrated_test_macro_f1:.4f}")
print(f"MAE: {calibrated_test_mae:.4f}")
print(f"Quadratic Weighted Kappa (QWK): {calibrated_test_quadratic_kappa:.4f}")
print(f"Spearman Correlation: {calibrated_test_spearman_rho:.4f}" if not np.isnan(calibrated_test_spearman_rho) else "Spearman Correlation: NaN")


# Per-Class Metrics (Precision, Recall, F1-Score)
print("\nPer-Class Metrics:")
per_class_df = pd.DataFrame({
    'Class': class_labels,
    'Precision': [calibrated_report[label]['precision'] for label in class_labels],
    'Recall': [calibrated_report[label]['recall'] for label in class_labels],
    'F1-Score': [calibrated_report[label]['f1-score'] for label in class_labels]
})
display(per_class_df)

# Confusion Matrix
print("\nConfusion Matrix:")
confusion_matrix_df = pd.DataFrame(calibrated_test_cm, index=class_labels, columns=class_labels)
display(confusion_matrix_df)

print("\n--- Reporting Complete ---")

# Optionally, save these final reported metrics to a file
report_summary_path = os.path.join(model_output_dir, 'calibrated_test_report_summary.txt')

with open(report_summary_path, 'w') as f:
    f.write("--- Reporting Metrics for Calibrated Test Set ---\n\n")
    f.write(f"Accuracy: {calibrated_test_accuracy:.4f}\n")
    f.write(f"Macro F1-Score: {calibrated_test_macro_f1:.4f}\n")
    f.write(f"MAE: {calibrated_test_mae:.4f}\n")
    f.write(f"Quadratic Weighted Kappa (QWK): {calibrated_test_quadratic_kappa:.4f}\n")
    f.write(f"Spearman Correlation: {calibrated_test_spearman_rho:.4f}" if not np.isnan(calibrated_test_spearman_rho) else "Spearman Correlation: NaN\n")
    f.write("\nPer-Class Metrics:\n")
    f.write(per_class_df.to_string(index=False) + "\n")
    f.write("\nConfusion Matrix:\n")
    f.write(confusion_matrix_df.to_string() + "\n")
    f.write("\n--- Reporting Complete ---\n")

print(f"\nReport summary saved to: {report_summary_path}")
```

## 4. Methodology / Proposed Approach

This section details the proposed hierarchical BERT model with attention pooling for ordinal movie script rating classification. We describe the data preprocessing pipeline, the novel model architecture, and the training strategy employed.

### Data Preprocessing

The raw movie script data, stored as text files, undergoes a series of preprocessing steps to prepare it for model input. Initially, filenames are parsed to extract metadata including the movie rating (G, PG, PG-13, R, NC-17), title, and year. A robust parsing function is implemented to handle variations in filename conventions. The extracted ratings are then mapped to numerical ordinal labels (0-4).

The script text itself is subjected to cleaning, including lowercasing, removal of punctuation and special characters, and normalization of whitespace. This results in a 'clean_script' for each movie.

Due to the potentially long nature of movie scripts, they are segmented into smaller, overlapping chunks to accommodate the input limitations of transformer models like BERT. Each script is split into chunks of a defined `MAX_CHUNK_SIZE` (set to 520 words, accounting for special tokens) with a specified `OVERLAP` (set to 50 words) between consecutive chunks. This overlap helps preserve context across chunk boundaries.

Finally, these text chunks are tokenized using the `bert-base-uncased` tokenizer, with truncation and padding applied to ensure uniform input length for the BERT encoder. The tokenized chunks, along with their corresponding attention masks and token type IDs, are stored for subsequent model input.

### Model Architecture

The proposed model, HierarchicalBERTOrdinal, employs a hierarchical approach to process the movie scripts. It consists of two main components: a chunk encoder and a document aggregator.

1.  **Chunk Encoder**: A `bert-base-uncased` model serves as the chunk encoder. Each individual tokenized script chunk is passed through this BERT model. The hidden state corresponding to the `[CLS]` token from the final layer of the BERT encoder is used as the fixed-size vector representation for that chunk.

2.  **Document Aggregator (Attention Pooling)**: Instead of a second BERT layer, a lightweight Attention Pooling mechanism is utilized to aggregate the representations of all chunks belonging to a single script. The sequence of chunk embeddings (one embedding per chunk) is fed into a simple linear layer that calculates an attention score for each chunk. A softmax function is applied to these scores to obtain attention weights, indicating the importance of each chunk to the overall document representation. The final document embedding is computed as a weighted sum of the chunk embeddings, where the weights are the attention scores. This approach allows the model to focus on the most salient parts of the script for rating prediction, offering a more computationally efficient alternative to a second BERT model.

3.  **Ordinal Regression Layer**: The aggregated document embedding is passed through a linear layer (`ordinal_layer`) with `num_labels - 1` output units. These outputs represent the logits for the cumulative probabilities of the ordinal classes. The model predicts the cumulative probability of a script belonging to class $k$ or higher, for $k$ from 1 to `num_labels - 1`.

The model is trained using a custom `OrdinalRegressionLoss` function, which is a Binary Cross-Entropy loss applied to the cumulative probabilities and the cumulative binary targets derived from the true ordinal labels.

### Training Strategy

The training strategy incorporates several techniques to improve model performance and address challenges such as class imbalance and the hierarchical nature of the data.

*   **Discriminative Learning Rates and Layer Freezing**: The lower layers of the BERT chunk encoder (specifically, the embedding layer and the first 6 transformer layers) are frozen during training by setting their `requires_grad` attribute to `False`. This preserves the general language understanding captured by these layers. The unfrozen BERT layers, the attention pooling layer, and the ordinal regression layer are trained with discriminative learning rates. A lower learning rate is used for the unfrozen BERT layers (`5e-5`), while a higher learning rate (`1e-4`) is applied to the newly added attention pooling and ordinal layers. This allows for fine-tuning of the pre-trained BERT layers while enabling faster learning for the task-specific layers. The AdamW optimizer is used for training.

*   **Weighted Random Sampling**: To mitigate the effects of class imbalance in the dataset, a `WeightedRandomSampler` is employed in the training dataloader. Sample weights are calculated based on the inverse frequency of each class in the training set, ensuring that instances from under-represented classes are sampled more frequently during training epochs.

*   **Learning Rate Scheduling and Gradient Clipping**: A linear learning rate scheduler with warmup is used. The learning rate increases linearly from 0 during a `warmup_steps` phase (set to 10% of total training steps) and then decays linearly to 0 over the remaining training steps. Gradient clipping with a maximum norm of 1.0 is applied to prevent exploding gradients during training.

*   **Early Stopping and Checkpoint Saving**: The training process includes early stopping based on the validation set's Quadratic Weighted Kappa (QWK). Training halts if the validation QWK does not improve for a specified number of `early_stopping_patience` epochs (set to 5). The model checkpoint with the best validation QWK is automatically loaded at the end of training.

*   **Post-hoc Ordinal Calibration**: After the training is complete, a post-hoc calibration step is performed. The optimal thresholds for converting the cumulative logits into discrete ordinal class predictions are determined by maximizing the Quadratic Weighted Kappa on the validation set predictions. These learned thresholds are then applied to the test set predictions to obtain the final calibrated predictions.

### Evaluation Metrics

The performance of the model is evaluated using a comprehensive set of metrics relevant to ordinal classification and general performance assessment:

*   **Accuracy**: The proportion of correctly predicted movie ratings.
*   **Mean Absolute Error (MAE)**: The average absolute difference between the true ordinal rating and the predicted ordinal rating.
*   **Mean Squared Error (MSE)**: The average squared difference between the true ordinal rating and the predicted ordinal rating.
*   **Linear and Quadratic Weighted Kappa (QWK)**: Cohen's Kappa score weighted linearly and quadratically, respectively. QWK is a common metric for ordinal tasks, penalizing misclassifications further away from the true label more heavily.
*   **Spearman's Rank Correlation Coefficient**: A non-parametric measure of the monotonic relationship between the true and predicted ordinal ratings.
*   **Cumulative Accuracy (Tolerance 0 and 1)**: The proportion of predictions that are exactly correct (Tol 0) or within one ordinal level of the true rating (Tol 1).
*   **Precision, Recall, and F1-Score**: Calculated per class to assess the model's performance on each individual rating category.
*   **Confusion Matrix**: A visualization showing the counts of true versus predicted ratings, providing insight into the model's misclassification patterns.
*   **ROC Curves and Precision-Recall Curves**: Generated per class to evaluate the model's ability to distinguish between classes based on the predicted probabilities.

The primary evaluation metric for model selection and early stopping is the **Quadratic Weighted Kappa (QWK)** on the validation set. Final performance is reported on the test set after applying post-hoc ordinal calibration.
